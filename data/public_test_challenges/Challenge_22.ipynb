# Problem setup:

A single photon is prepared in a coherent superposition of two spatial modes: mode $A$, which passes through the laboratory of a sender (Alice), and a reference mode $R$, which travels directly to the receiver (Bob) and is never touched by Alice. In both modes the photon is horizontally polarized,
$$
|\Psi_\theta\rangle=\cos\theta\,|A,H\rangle+\sin\theta\,|R,H\rangle,\qquad \theta\in[0,\pi/2].
$$

To send message $x$, which occurs with prior probability $p_x$, Alice acts on mode $A$ only. She passes it through an absorber that removes the photon with probability $\gamma_x\in[0,1]$, i.e. a loss (amplitude-damping) channel with transmissivity $1-\gamma_x$, so the transmitted amplitude is $\sqrt{1-\gamma_x}$ and $\gamma_x=1$ is complete blocking. She also applies a polarization unitary $U_x\in\mathrm{U}(2)$ (its global phase is a phase shift of mode $A$). Bob receives both modes. Write $|\mathrm{vac}\rangle$ for "no photon arrives" and $|u_x\rangle=U_x|H\rangle=\alpha_x|H\rangle+\beta_x|V\rangle$ with $|\alpha_x|^2+|\beta_x|^2=1$. Bob's state for message $x$ is
$$
\rho_x=\gamma_x\cos^2\theta\,|\mathrm{vac}\rangle\langle\mathrm{vac}|+|\psi_x\rangle\langle\psi_x|,\qquad
|\psi_x\rangle=\sqrt{1-\gamma_x}\,\cos\theta\,|A\rangle\otimes|u_x\rangle+\sin\theta\,|R,H\rangle .
$$
In the ordered basis $\{|\mathrm{vac}\rangle,|A,H\rangle,|A,V\rangle,|R,H\rangle\}$, with $c=\cos\theta$ and $s=\sin\theta$,
\begin{equation}
\rho_x=\left(\begin{array}{cccc}
\gamma_x c^2 & 0 & 0 & 0\\
0 & (1-\gamma_x)c^2|\alpha_x|^2 & (1-\gamma_x)c^2\alpha_x\beta_x^* & \sqrt{1-\gamma_x}\,cs\,\alpha_x\\
0 & (1-\gamma_x)c^2\alpha_x^*\beta_x & (1-\gamma_x)c^2|\beta_x|^2 & \sqrt{1-\gamma_x}\,cs\,\beta_x\\
0 & \sqrt{1-\gamma_x}\,cs\,\alpha_x^* & \sqrt{1-\gamma_x}\,cs\,\beta_x^* & s^2
\end{array}\right).
\end{equation}

The classical-quantum state is $\sigma_{XB}=\sum_x p_x|x\rangle\langle x|\otimes\rho_x$, and its Holevo information is
$$
\chi=S\Big(\sum_x p_x\rho_x\Big)-\sum_x p_x\,S(\rho_x),\qquad S(\rho)=-\mathrm{Tr}\,\rho\log_2\rho .
$$
All entropies are in bits. $h_2(z)=-z\log_2 z-(1-z)\log_2(1-z)$ is the binary entropy, with $0\log_2 0=0$.
