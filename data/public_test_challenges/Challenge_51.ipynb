# Problem setup:
Consider the following discrete time process on the one-dimensional integer lattice. At time $0$ there is one composite particle at the origin. At any later time the state is either a single composite particle at some site $x$, or two distinguishable constituent particles at distinct sites $x_1 \neq x_2$.

If the state at time $s$ is a composite particle at site $x$, then the next time step is one of the following:

1. the composite particle moves to $x+1$ or $x-1$; this step has weight $\lambda$, with $\lambda > 0$;
2.  the composite particle splits into two constituents occupying the adjacent sites $x-1$ and $x+1$. There are exactly $g$ distinguishable split outcomes, where $g \geq 2$ is an integer. The number $g$ includes the assignment of the constituent labels to the left and right sites, so there is no additional factor of $2$ at a splitting step. For example, if the constituents are red and blue, then $g=2$ because red-left/blue-right and blue-left/red-right are the two split outcomes. A split step has unit weight apart from this multiplicity.

If the state at time $s$ consists of two constituents, then

1.  each constituent independently moves one site left or right in the next time step, and each constituent-particle step has unit weight.
2.  if the two constituents occupy the same lattice site after the simultaneous move, they immediately recombine into a composite particle at that site.

Constituents never split.  

Let $Z(0)=1$. For $t\geq 1$, let $Z(t)$ be the total weight of all duration-$t$ histories whose final move, from time $t-1$ to time $t$, is a recombination event. Different choices of composite moves, constituent moves, and split outcomes are counted as different configurations. Equivalently, after summing over the explicit labels of the $g$ split outcomes, a history with $N_{\rm comp}$ composite-particle move steps and $N_{\rm split}$ splitting steps contributes
\begin{equation}
g^{N_{\rm split}}\lambda^{N_{\rm comp}} .
\end{equation}
