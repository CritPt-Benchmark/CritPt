$$
\alpha_i^N=1,\qquad
\alpha_i^\dagger=\alpha_i^{-1},\qquad
\alpha_j\alpha_k=e^{[2\pi i{\rm sgn}(j-k)]/N}\alpha_k\alpha_j.
$$

Consider the four-stage adiabatic tunneling cycle

$$
H_{34}\rightarrow H_{23}\rightarrow H_{12}\rightarrow H_{13}\rightarrow H_{34},
$$

where

$$
H_{ij}=-t\left(e^{-i\phi_{ij}/N}\alpha_i^\dagger\alpha_j+H.c.\right),\qquad t>0.
$$

Here $\phi_{ij}$ is a phase between sites $i$ and $j$, and $H.c.$ stands for Hermitian conjugate.

Each arrow $H_A\rightarrow H_B$ denotes the adiabatic interpolation

$$
H(s)=(1-s)H_A+sH_B,\qquad 0\le s\le1.
$$

Only these two couplings are present during each interpolation, and all tunneling phases $\phi_{ij}$ are held constant throughout the cycle. Assume that the ground-state manifold remains degenerate and separated from excited states by a nonzero gap, with no additional ground-state level crossings.
